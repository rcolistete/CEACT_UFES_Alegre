## Estruturas de Teste em Python : if, else, elif

Condições de teste fixas usando valores verdadeiro (`True`) e falso (`False`) :

```python
if True:
    print("Como o 1o teste é sempre verdadeiro, esse print é sempre exibido...")
```

```python
if False:
    print("Como o 1o teste é sempre falso, esse print nunca é exibido...")
else:    
    print("Mas esse 2o print, do else/senão, é sempre exibido...")
```

```python
if False:
    print("Como o 1o teste é sempre falso, esse 1o print nunca é exibido...")
elif True:
    print("Como o 2o teste é sempre verdadeiro, esse print é sempre exibido...")    
else:    
    print("Mas esse 3o print, do else/senão, nunca é exibido...")
```

Condições de teste com comparações :

```python
acel = 42 
if acel > 0:
    print("Aceleração é positiva")
elif acel == 0:
    print("Aceleração é nula")
else:    
    print("Aceleração é negativa")
```

```python
acel = -7
if acel > 0:
    print("Aceleração é positiva")
elif acel == 0:
    print("Aceleração é nula")
else:    
    print("Aceleração é negativa")
```

```python
acel = 0
if acel > 0:
    print("Aceleração é positiva")
elif acel == 0:
    print("Aceleração é nula")
else:    
    print("Aceleração é negativa")
```

Condições de teste com combinações de comparações com `and`, `or` e `not` :

```python
def teste_massa(m):
    if (m >= 10) and (m <= 100):
        print("m entre 10 e 100")
    elif (m < 0) or (m > 100) :
        print("m é negativo ou maior que 100")
    else:    
        print("m maior que 0 e menor que 10")
```

```python
teste_massa(42)
```

```python
teste_massa(200)
```

```python
teste_massa(4)
```

```python
def teste_idade(idade):
    if not (idade >= 0):
        print("Idade não pode ser negativa")
    elif (m < 18) :
        print("Com ", idade, " anos, você é menor de idade")
    else:    
        print("Com ", idade, " anos, você é maior de idade")
```

```python
teste_idade(-42)
```

```python
teste_idade(13)
```

```python
teste_idade(26)
```

### Símbolos e Expressões em SymPy

Todo símbolo (variável literal, sem valor associado) deve ser declarado dessa forma em SymPy, para diferenciar dos  objetos em Python que precisam ter valores antes de serem usados :

```python
x, y, z, t, a, b, c = symbols('x, y, z, t, a, b, c')
```

```python
x
```

```python
?x
```

```python
(x + 2)**2
```

```python
expand((x + 2)**2)
```

```python
expand((x + 2)**12)
```

```python
expr1 = expand((x + 5)**3); expr1
```

```python
factor(expr1)
```

```python
expr2 = expand(9*(x + y)**3); expr2
```

```python
simplify(expr2)
```
