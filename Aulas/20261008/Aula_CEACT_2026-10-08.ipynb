## Oscilador Amortecido

Um corpo tem seu movimento descrito pela função posição :

$$x(t) = A \exp\left(\frac{-b\,t}{2\, m}\right) \cos(\omega\,t)$$

Considere $A = 3,0\,m$, $b = 2,0\,kg/s$, $m = 25\,kg$ e $\omega = \pi\,rad/s$ :

### 1) Gráfico de $x(t)$

```python
from sympy import *
```

```python
A, b, omega, m, t = symbols('A, b, omega, m, t')
```

```python
def posicaox(A, b, m, omega, t):
    return A*exp(-b*t/(2*m))*cos(omega*t)
```

```python
posicaox(A, b, m, omega, t)
```

```python
posicaox(3, 2, 25, pi, t)
```

```python
posicaox(3, 2, 25, pi, 5.0)
```

```python
posicaox(3, 2, 25, pi, 70.0)
```

```python
plot(posicaox(3, 2, 25, pi, t), (t, 0, 30));
```

A gráfico da posição mostra que a amplitude inicial é $3,0\,m$, porém vai caindo com o tempo no oscilador amortecido.

```python
plot(posicaox(3, 2, 25, pi, t), (t, 0, 33), 
  xlabel='t(s)', ylabel='x(m)', 
  title='Posição do Oscilador Amortecido');
```

Com $b$ maior, a oscilação fica mais amortecida :

```python
plot(posicaox(3, 5, 25, pi, t), (t, 0, 33), 
  xlabel='t(s)', ylabel='x(m)', 
  title='Posição do Oscilador Amortecido');
```

Aumentando $m$ (massa), diminui o amortecimento da oscilação :

```python
plot(posicaox(3, 2, 100, pi, t), (t, 0, 33), 
  xlabel='t(s)', ylabel='x(m)', 
  title='Posição do Oscilador Amortecido');
```

```python
plot(posicaox(3, 2, 5, pi, t), (t, 0, 33), 
  xlabel='t(s)', ylabel='x(m)', 
  title='Posição do Oscilador Amortecido');
```

Aumentando/diminuindo $\omega$ (frequência angular), aumenta/diminui o número de oscilações por período de tempo :

```python
plot(posicaox(3, 2, 100, 2*pi, t), (t, 0, 33), 
  xlabel='t(s)', ylabel='x(m)', 
  title='Posição do Oscilador Amortecido');
```

```python
plot(posicaox(3, 2, 100, pi/2, t), (t, 0, 33), 
  xlabel='t(s)', ylabel='x(m)', 
  title='Posição do Oscilador Amortecido');
```

### 2) Velocidade $v(t)$

$$ v(t) = \frac{d\,x(t)}{d\,t} $$

`diff()` com só 2 argumentos faz derivada de 1a ordem em relação à variável do 2o argumento :

```python
diff(posicaox(A, b, m, omega, t), t)
```

Aqui uma derivada de 4a ordem :

```python
diff(posicaox(A, b, m, omega, t), t, 4)
```

Definindo função velocidade em termos da derivada 1a da posição em relação ao tempo. Ao chamar a função o parâmetro `t` tem que ser simbólico (não pode ser numérico) :

```python
def velocidade(A, b, m, omega, t):
    return diff(posicaox(A, b, m, omega, t), t)
```

```python
velocidade(A, b, m, omega, t)
```

```python
velocidade(3, 2, 25, pi, t)
```

```python
plot(velocidade(3, 2, 25, pi, t), (t, 0, 33), 
  xlabel='t(s)', ylabel='v(m/s)', 
  title='Velocidade do Oscilador Amortecido');
```

Com $b$ maior, a oscilação da velocidade também fica mais amortecida :

```python
plot(velocidade(3, 5, 25, pi, t), (t, 0, 33), 
  xlabel='t(s)', ylabel='v(m/s)', 
  title='Velocidade do Oscilador Amortecido');
```

Aumentando $m$ (massa), diminui também o amortecimento da oscilação da velocidade :

```python
plot(velocidade(3, 2, 100, pi, t), (t, 0, 33), 
  xlabel='t(s)', ylabel='v(m/s)', 
  title='Velocidade do Oscilador Amortecido');
```

Aumentando/diminuindo $\omega$ (frequência angular), aumenta/diminui o número de oscilações por período de tempo :

```python
plot(velocidade(3, 2, 25, 2*pi, t), (t, 0, 33), 
  xlabel='t(s)', ylabel='v(m/s)', 
  title='Velocidade do Oscilador Amortecido');
```

```python
plot(velocidade(3, 2, 25, pi/2, t), (t, 0, 33), 
  xlabel='t(s)', ylabel='v(m/s)', 
  title='Velocidade do Oscilador Amortecido');
```

### 3) Aceleração $a(t)$

$$ a(t) = \frac{d\,v(t)}{d\,t} $$

$$ a(t) = \frac{d^2\,x(t)}{d\,t^2} $$

```python
diff(posicaox(A, b, m, omega, t), t, 2)
```

Definindo função aceleração em termos da derivada 2a da posição em relação ao tempo. Ao chamar a função o parâmetro `t` tem que ser simbólico (não pode ser numérico) :

```python
def aceleracao(A, b, m, omega, t):
    return diff(posicaox(A,b,m,omega,t), t, 2)
```

```python
aceleracao(A, b, m, omega, t)
```

```python
aceleracao(3, 2, 25, pi, t)
```

Notar que pode ser usado $\TeX$ nos gráficos :

Com $b$ maior, a oscilação da aceleração também fica mais amortecida :

```python
plot(aceleracao(3, 5, 25, pi, t), (t, 0, 33), 
  xlabel='t(s)', ylabel='$a(m/s^2)$', 
  title='Aceleração do Oscilador Amortecido');
```

### 4) Energia cinética $K$

$$ K = \frac{m v^2}{2} $$

A energia cinética $K$ :

```python
m*(diff(posicaox(A, b, m, omega, t), t)**2)/2
```

```python
expand(m*(diff(posicaox(A, b, m, omega, t), t)**2)/2)
```

```python
simplify(m*(diff(posicaox(A, b, m, omega, t), t)**2)/2)
```

Definindo função energia cinética. Ao chamar a função o parâmetro `t` tem que ser simbólico (não pode ser numérico) :

```python
def energiacineticaK(A, b, m, omega, t):
    return m*(velocidade(A, b, m, omega, t)**2)/2
```

```python
plot(energiacineticaK(3, 2, 25, pi, t), (t, 0, 33),
     xlabel='t', ylabel='K',
     title=u'Energia cinética do oscilador amortecido');
```

Com $b$ maior, a oscilação da energia cinética também fica mais amortecida :

```python
plot(energiacineticaK(3, 5, 25, pi, t), (t, 0, 33),
     xlabel='t', ylabel='K',
     title=u'Energia cinética do oscilador amortecido');
```

Aumentando $m$ (massa), diminui também o amortecimento da oscilação da energia cinética :

```python
plot(energiacineticaK(3, 2, 100, pi, t), (t, 0, 33),
     xlabel='t', ylabel='K',
     title=u'Energia cinética do oscilador amortecido');
```

Aumentando/diminuindo $\omega$ (frequência angular), aumenta/diminui o número de oscilações por período de tempo :

```python
plot(energiacineticaK(3, 2, 25, 2*pi, t), (t, 0, 33),
     xlabel='t', ylabel='K',
     title=u'Energia cinética do oscilador amortecido');
```

```python
plot(energiacineticaK(3, 2, 25, pi/2, t), (t, 0, 33),
     xlabel='t', ylabel='K',
     title=u'Energia cinética do oscilador amortecido');
```

A energia cinética é sempre positiva e começa nula, oscila e sua amplitude diminui com o tempo.

### Vetores em Python com NumPy para vetores numéricos

Com NumPy temos "array" que permite definir vetores (e matrizes, etc) matemáticos, incluindo produto escalar, etc. A limitação é que só para valores numéricos.

```python
import numpy as np
```

A função "array" do NumPy define um vetor numérico :

```python
vetor2 = np.array([1, 2, 3])
```

```python
vetor2[0]
```

```python
np.dot(vetor2, vetor3)
```

```python
np.cross(vetor2, vetor3)
```

```python
np.cross(vetor2, vetor2)
```

### Vetores em Python com SymPy para vetores numéricos e simbólicos

```python
from sympy import *
```

Função "Matrix" define um vetor (que é um caso específico de uma matriz, aqui um vetor coluna 3x1) :

```python
vetor4 = Matrix(3,1,[1,2,3])
```

```python
vetor4
```

```python
init_printing()
```

```python
vetor4
```

"dot" é um método do vetor para calcular o produto escalar :

```python
vetor4.dot(vetor4)
```

"cross" é um método do vetor para calcular o produto vetorial :

```python
vetor4.cross(vetor4)
```

Vetores com elementos simbólicos :

```python
t, R, omega = symbols("t, R, omega")
```

Vetor posição de um movimento circular uniforme em 2D :

```python
r = Matrix(2,1,[R*cos(omega*t),R*sin(omega*t)])
```

```python
r
```

```python
r.dot(r)
```

```python
simplify(r.dot(r))
```

Vetor velocidade é a derivada do vetor posição em relação ao tempo t, onde "diff" é um método do vetor :

```python
v = r.diff(t); v
```

```python
simplify(v.dot(v))
```

Vetor posição r é ortogonal ao vetor velocidade v :

```python
r.dot(v)
```

Vetor aceleração é a derivada do vetor velocidade em relação ao tempo t :

```python
a = v.diff(t); a
```

```python
simplify(a.dot(a))
```

Ou seja, $a = v^2/R$ :

```python
simplify(v.dot(v))/R
```

Vetor velocidade v é ortogonal ao vetor aceleração a :

```python
v.dot(a)
```
