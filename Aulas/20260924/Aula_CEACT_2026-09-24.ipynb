## Estruturas de Teste em Python : if, else, elif

Condições de teste fixas usando valores lógicos verdadeiro (`True`) e falso (`False`) :

```python
if True:
    print("Como o 1o teste é sempre verdadeiro, esse print é sempre exibido...")
```

```python
if False:
    print("Como o 1o teste é sempre falso, esse print nunca é exibido...")
else:    
    print("Mas esse 2o print, do else/senão, é sempre exibido...")
```

```python
if False:
    print("Como o 1o teste é sempre falso, esse 1o print nunca é exibido...")
elif True:
    print("Como o 2o teste é sempre verdadeiro, esse print é sempre exibido...")    
else:    
    print("Mas esse 3o print, do else/senão, nunca é exibido...")
```

Condições de teste com comparações :

```python
acel = 42 
if acel > 0:
    print("Aceleração é positiva")
elif acel == 0:
    print("Aceleração é nula")
else:    
    print("Aceleração é negativa")
```

Condições de teste com combinações de comparações com operadores lógicos `and`, `or` e `not` :

```python
def teste_massa(m):
    if (m >= 10) and (m <= 100):
        print("m entre 10 e 100")
    elif (m < 0) or (m > 100) :
        print("m é negativo ou maior que 100")
    else:    
        print("m maior que 0 e menor que 10")
```

```python
teste_massa(42)
```

```python
teste_massa(200)
```

```python
teste_massa(4)
```

```python
def teste_idade(idade):
    if not (idade >= 0):
        print("Idade não pode ser negativa")
    elif (idade < 18) :
        print("Com ", idade, " anos, você é menor de idade")
    else:    
        print("Com ", idade, " anos, você é maior de idade")
```

```python
teste_idade(-42)
```

```python
teste_idade(13)
```

```python
teste_idade(26)
```

Entrada de dados via `input()` :

```python
def teste_42():
    n = int(input("Insira um número inteiro: "))
    if n != 42:
        print("O número ", n, "não é a resposta para a vida, o universo e tudo mais")
    else:
        print("O número ", n, "é a resposta para a vida, o universo e tudo mais")
```

```python
teste_42()
```

```python
teste_42()
```

## Estruturas de Repetição em Python : `while`

```python
a = 0
while a < 10:
    print(a)
    a = a + 1
```

```python
a = 0
while a < 10:
    print("a = ", a, " e a ao quadrado = ", a**2)
    a += 1
```

```python
v = [1, 9, 4, 2]
```

```python
i = 0
while i < len(v):
    print("índice=", i, "elemento de v :", v[i])
    i += 1
```
