```python
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib
```

```python
flights_data = sns.load_dataset("flights")
flights_data.head()
```

```python
sns.scatterplot(data=flights_data, x="year", y="passengers");
```

```python
sns.lineplot(data=flights_data, x="year", y="passengers");
```

```python
sns.barplot(data=flights_data, x="year", y="passengers");
```

```python
import seaborn as sns
from matplotlib import pyplot as plt
df = sns.load_dataset('iris')
sns.set_style("ticks")
sns.pairplot(df,hue = 'species',diag_kind = "kde",kind = "scatter",palette = "husl")
plt.show()
```

# [Bokeh - bokeh.org](https://bokeh.org/)

Vide [Bokeh Quickstart](https://nbviewer.jupyter.org/github/bokeh/bokeh-notebooks/blob/master/quickstart/quickstart.ipynb) :

```python
from bokeh.plotting import figure 
from bokeh.io import output_notebook, show
```

```python
output_notebook()
```

```python
from numpy import cos, linspace
x = linspace(-6, 6, 100)
y = cos(x)
```

```python
p = figure(width=500, height=500)
p.circle(x, y, size=7, color="firebrick", alpha=0.5)
```

```python
show(p)
```
